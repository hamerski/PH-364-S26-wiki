Pandas provides a set of tools for working with datasets that are not immediately read for analysis. These include methods such as reading files that contain metadata, selecting only the needed columns, and sorting the data into a usable order. Pandas also can sample rows based on time, such as choosing the entries closet to specific dates or intervals. These features make it possible to turn raw data into a structured table that can be used for analysis.

# Reading and Organzing Data

The main Pandas functions used were to read the files, organize the columns, and sample the data at regular time intervals. read_csv, is used with skips rows and nrows so that only the actual data block is read and the metadata above and below it is ignored. After loading, the columns are renamed using df.columns, because the original Horizons files does not provide clean names. For example: 

```python
df_mercury = pd.read_csv('mercury-orbit-200yrs.csv', skiprows = 47, nrows = 73048)
df_mercury.columns = ['jd', 'calendar_date', 'x', 'y', 'z', 'vx', 'vy', 'vz', 'LT', 'R', 'RR', 'extra']
```

skipsrows=47 removes the first 47 lines of metadata so the read begins at the first actual data row. nrows=73048 tells pandas to read the next 73,048 rows and stop, which stops it from pulling in the footer text after the data block. After loading, the columns are replaced by assigning a list to df.columns, which gives each column a clear label like x, y, and z. This makes the position values easy to accces and avoids dealing with other names given in the data that were unreadable.

Pandas was needed in the project because the raw Horizons file could not be used directly. The file included a lot of metadata in the beginning and some more information at the end, and columns that were not easy to read, so the data had to be organized before any analysis could happen. Using Pandas made it straightfoward to load only the data block, assign readable column names, and access the needed postion values needed to animate and analyze 200 years of planetary motion each day.
