# FuncAnimation Review

FuncAnimation is a useful module in the matplot lib library. This module can help people visualize models, or other desired code by animating it however you'd like. For Computational Physics, when modeling physical problems, it is often times useful to see how the situation is changing with respect to time or other variables. That is where the FuncAnimation module can be of use.

First lets look at some of the matplotlib information regarding FuncAnimation:



To implement this code you must first import the module, to do this you would use:

```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
```

This gives our code access to all that FuncAnimation has to offer. Next you would create the code for the situation you wanted to model, usually something with a variable you would like to change, and observe how the situation changes due to that variable. Then with the FuncAnimation function, it repeatedly calls the animation function, updating the figure each time. Example:

```python
import numpy as np

#Animation
plt.rcParams["animation.html"] = "jshtml"
plt.rcParams['figure.dpi'] = 60
fig, ax = plt.subplots()

t = np.arange(0,10,0.1)
pos = np.array([np.copy(t), np.copy(t)])

def animate(i):
    ax.clear()

    plt.xlabel("X-Position (m)")
    plt.ylabel("Y-Position (m)")
    plt.title("Y-Position vs. X-Position")
    plt.ylim(0, np.max(pos[1])*1.25)
    plt.xlim(0, np.max(pos[0])*1.25)
    plt.grid()
    
    #This type of animation reveals the plot bit by bit instead of recalculating everything.
    plt.plot(pos[0][:i+1], pos[1][:i+1])

anim = FuncAnimation(fig, animate, frames=range(0, len(t), 5))
anim
```

In the code above, I instead chose to reveal the rest of the graph bit by bit instead of creating the change as animation runs. I created a simple linear graph, and using the argument I put in animation, the "i," I plotted the next iteration of "i." This animation can also be used like I stated above to actively change initial data, and this would require putting that change function into the animation function anad calling it how you'd like. This first section,

```python
plt.rcParams["animation.html"] = "jshtml"
plt.rcParams['figure.dpi'] = 60
fig, ax = plt.subplots()
```

simply sets up the figure, axis, and the animation. It tells Jupyter how to display the animation first, then the resolution/ size of the displayed animation, and finally the plots to put the animation onto. The in-between stuff:

```python
t = np.arange(0,10,0.1)
pos = np.array([np.copy(t), np.copy(t)])

def animate(i):
    ax.clear()

    plt.xlabel("X-Position (m)")
    plt.ylabel("Y-Position (m)")
    plt.title("Y-Position vs. X-Position")
    plt.ylim(0, np.max(pos[1])*1.25)
    plt.xlim(0, np.max(pos[0])*1.25)
    plt.grid()
    
    #This type of animation reveals the plot bit by bit instead of recalculating everything.
    plt.plot(pos[0][:i+1], pos[1][:i+1])
```

is just me setting up the situation, but the important parts are clearing your ax each time so that the animation gets rid of the previous graph, and updates it with the freshly changed/ revealed one next, then I customize my graphs how I'd like with labels and stuff, and finally call the change funciton, or in this case, the "reveal" function. The last step,

```python
anim = FuncAnimation(fig, animate, frames=range(0, len(t), 5))
anim
```

involves storing the animation, and then displaying it later. The arguments inside of FuncAnimation are shown in the first documentation I took a screenshot of above, but basically, the first argument is saying we want to display this figure, the second is saying this is the funciton we want to keep calling, and the third argument is the argument for animation, "i" in this case, which helps determine how many times to run the animation function, and in our case "how large" of a step to take since we are displaying in intervals of 5 for i.
