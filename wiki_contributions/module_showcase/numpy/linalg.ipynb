    For this part of the project I chose to examine numpy libraries and a specfic function within said library. This function is np.linalg.eig(b). np.linalg.eig(b) computes the eigen values and vectors of a 2d array b. b must be a square 2D array (meaning the rows equal the columns) so that we an commute this and if it is not a sqaure array you will get an error. If the eigenvalue math does not converge then we should use np.linalg.LinAlgError(b) however, in our final project we did not need this. Now the function will return two different 1D arrays one for the eigenvalues and one for the eigenvectors. Looking first a the 1D array for eigenvalues the a not necessarily order and the the array can be complex with both the imaginary and real part. Then the eigenfunction array will be the normalized eigenvetors and the eigenvector placement should correspond with the eigenvalue associated with it. For example eigenvectors[:,i] will go with eigenvalues[i]. You should use this when you need eigenvalues and vectors as matrix multpication in python (or on paper) can get difficult. Below is an example of how this can be implemented.

```python
import numpy as np

b = np.array([[-3, 4, 0, 7],
             [3, 6, 2, 0],
             [0, 1, 9, 1],
             [9, 0, 8, -2]])

print()
print(np.linalg.eig(b))
print()
```
