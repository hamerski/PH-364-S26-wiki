A coding tool that we used was scipys special functions scipy.special, and more specifically we used the .genlaguerre() and .sph_harm_y() functions which calculate laguerre polynomials nd spherical harmonics for given quantum and angular variables. You use these functions when you need to quickly calculate the spherical harmonics or laguerre polynomials for a certain set of variables, especialy if those variables are constantly changing. 

```python
import scipy.special as sp

#Parameters
a = 2 #Must be greater then -1
n = 1 #degree of the polynomial

laguerre_polynomial = sp.genlaguerre(n, a)
print(laguerre_polynomial)
```
